```python
import spacy
from nltk import ngrams
nlp=spacy.load("en_core_web_sm")
doc=nlp("ai is simulation of human intelligence,by machine learning.it includes excepts system nlp speech recognization and machine vision")
bigrams=ngrams(doc,2)
print(list(bigrams))
```

```python
import spacy
from nltk import trigrams
nlp=spacy.load("en_core_web_sm")
doc=nlp("ai is simulation of human intelligence,by machine learning.it includes excepts system nlp speech recognization and machine vision")
trigrams=trigrams(doc)
print(list(trigrams))
```
